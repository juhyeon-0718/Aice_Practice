```python
# 범위 지정
top_10_rows = data[:10]
top_10_rows
```

```python
# 특정 행, 열 선택 : iloc 또는 loc 사용
## iloc : 인덱스 기준 데이터 선택 
value = data.iloc[0,1] # 첫 번째 행의 두 번째 열 데이터 
print(value)
```

```python
## loc : 행과 열 이름을 기준으로 데이터 선택
value = data.loc[0, 'quality'] # 첫 번째 행의 'quality'열의 값 선택
print(value)
```

```python
# 데이터 조건으로 검색

## 단일 조건 
### ex) 품지링 7이상인 데이터 선택
high_quality = data[data['quality'] >= 7]
print(high_quality.head())
```

```python
## 다중 조건
### ex) 품질이 8이상이고 pH가 3이상인 데이터 선택
high_quality_alcohol = data[(data['quality'] >= 8) & (data['pH'] >= 3)]
print(high_quality_alcohol.head())
```
