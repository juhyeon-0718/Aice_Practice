```python
# 데이터 정렬
import pandas as pd
path = 'C:/juehyeon/aice_test/materials/datasets/'
df = pd.read_csv(path + 'heart.csv')
```

```python
## 특정 열 기준으로 정렬 sort_values(by = 기준 열, ascending = True)
df_sorted_by_age = df.sort_values(by = 'age', ascending = True)
print(df_sorted_by_age)
```

```python
## 여러 열 기준 정렬 sort_values(by = ['기준열1', '기준열2], ascending = ['오름차순', '내림차순'])
df_sorted_by_multiple = df.sort_values(by = ['age', 'thalachh'], ascending = [True, False])
df_sorted_by_multiple.head()
```

```python
## 인덱스 기준 정렬 : sort_index()
df_sorted_by_index = df.sort_index(ascending = False)
df_sorted_by_index.head()
```

```python
### 인덱스를 특정 열로 설정
data = {'Name' : ['Alice', 'Bob', 'Charlie'],
        'Age' : [25, 30, 35]}
df_n = pd.DataFrame(data)
df_n.set_index('Name', inplace = True)
print(df_n)
```

```python
### 인덱스를 기본값으로 재설정
df_n.reset_index(inplace = True)
print(df_n)
```

```python
## 열 이름 기준으로 정렬 : sort_index(axis = 1)
df_sorted_by_columns = df.sort_index(axis = 1, ascending = True)
print(df_sorted_by_columns.head())
```
