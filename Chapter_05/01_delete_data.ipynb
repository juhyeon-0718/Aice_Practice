```python
# 데이터 삭제 
## 특정 행만 삭제하기 df.drop(index = x)
```

```python
import pandas as pd

data = {'Name' : ['Alice', 'Bob', 'Charlie', 'David'],
        'Age' : [25, 30, 35, 40],
        'Score' : [90, 85, 75, 60]}
df = pd.DataFrame(data)

df_dropped = df.drop(index = 2)
print(df_dropped)
```

```python
## 특정 열만 삭제하기 drop( colummns = ['컬럼명'] ) 
df_dropped = df.drop(columns = ['Score'])
print(df_dropped)
```

```python
## 결측치가 많은 행 삭제 dropna()
data_with_na = {'Name' : ['Alice', 'Bob', 'Charlie', 'David'],
                'Age' : [25, None, 35, 40],
                'Score' : [90, 85, None, 60]}
df_with_na = pd.DataFrame(data_with_na)
```

```python
df_cleaned = df_with_na.dropna()
print(df_cleaned)
```

```python
## 중복 데이터 삭제 : drop_duplicates()
data_with_duplicates = {'Name' : ['Alice', 'Bob', 'Alice', 'David'],
                        'Age' : [25, 30, 25, 40],
                        'Score' : [90, 85, 90, 60]}
df_with_duplicates = pd.DataFrame(data_with_duplicates)
df_unique = df_with_duplicates.drop_duplicates()
print(df_unique)
```
