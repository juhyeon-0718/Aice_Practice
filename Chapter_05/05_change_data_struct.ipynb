```python
import pandas as pd
path = 'C:/juehyeon/aice_test/materials/datasets/'
df = pd.read_csv(path + 'heart.csv')
```

```python
df.columns = ['나이', '성별', '흉통유형', '안정시혈압', '콜레스테롤', '공복혈당', '심전도', '최대 심박수',
              '운동유발협심증', 'ST 우울증', '운동후 ST경사', '주요혈관수', '결함유형', '심장질환여부']
```

```python
# 그룹화와 집계 
## 성별을 기준으로 데이터 그룹화
groupped= df.groupby('성별')
groupped.size()
```

```python
## 성별별 주요 열의 평균 집계
groupped_mean = groupped.mean()
print(groupped_mean)
```

```python
## 여러 집계 함수 사용 : agg()
### agg() : 여러 열에 다양한 집계 함수 동시 적용, 딕셔너리 형태로 열 이름과 함수 목록 지정
groupped_stats = groupped.agg({
    '나이' : ['max' , 'min'],
    '콜레스테롤' : ['mean', 'std']
})
print(groupped_stats)
```

```python
## 다중 그룹화
groupped_multi = df.groupby(['성별', '흉통유형']).mean()
groupped_multi
```

```python
# Pivot Table 조작 : 특정 기준에 따라 데이터를 요약하거나 변환하는데 사용하는 도구
## 1. pivot table 생성
pivot = df.pivot_table(index = '성별', columns = '흉통유형', values = '최대 심박수', aggfunc = 'mean')
### 해석 ) 성별을 인덱스로 지정, 흉통유형을 열로 지정, 최대 심박수 값을 피벗 테이블에서 요약 데이터로 사용, 최대 심박수의 평균값 계산
pivot
```

```python
## 2. 집계함수 적용
pivot = df.pivot_table(index = '성별', columns = '흉통유형', values = '최대 심박수', aggfunc = ['mean', 'max'])
pivot
```

```python
## 3. 다중 인덱스 피벗 테이블
pivot_multi_index = df.pivot_table(index = ['성별', '흉통유형'] , values = '최대 심박수', aggfunc = 'mean')
pivot_multi_index
```

```python
'''
Pivot Table 주요 옵션
1) index : 행 인덱스로 설정할 열
2) columns : 열로 설정할 데이터
3) values : 요약할 데이터
4) aggfunc : 집계 함수 지정
5) fill_value : 결측값을 대체할 값 
'''
```

```python
## 4. pivot table 결측값 처리
data = {
    '성별' : ['남성', '여성', '남성', '여성'],
    '흉통유형' : [1,2,3,1],
    '최대 심박수' : [150, 160, 155, 145]
}
df = pd.DataFrame(data)
print('\n 원본 데이터 프레임 \n' ,df)

### pivot table 생성
pivot = df.pivot_table(index = '성별', columns = '흉통유형', values = '최대 심박수', aggfunc = 'mean')
print('\n 원본 pivot table \n' ,pivot)
```

```python
### 결측값을 특정값으로 대체하기
pivot_filled = df.pivot_table(index = '성별' , columns = '흉통유형', values = '최대 심박수', aggfunc = 'mean', fill_value = 0)
print('결측값을 0으로 대체 \n' , pivot_filled)
```

```python
### 평균값으로 대체 
pivot_mean_filled =  pivot.apply(lambda x: x.fillna(x.mean()), axis = 0)
pivot_mean_filled
```

```python
### 특정값으로 설정
pivot_setting_value = pivot.fillna(999)
pivot_setting_value 
```

```python
# 인덱스와 컬럼 계층 변경
## 인덱스 계층 : 여러 개의 인덱스 레벨을 사용해 데이터를 계층적으로 관리하는 방식 
## 컬럼 계층 : 여러 개의 컬럼 레벨을 사용해 데이터를 다차원적으로 표현하는 방식
df.columns = ['나이', '성별', '흉통유형', '안정시혈압', '콜레스테롤', '공복혈당', '심전도', '최대 심박수',
              '운동유발협심증', 'ST 우울증', '운동후 ST경사', '주요혈관수', '결함유형', '심장질환여부']
```

```python
## 1. 그룹화 후 다중 인덱스 생성
groupped = df.groupby(['성별', '흉통유형']).mean()
groupped
```

```python
## 2. 다중 인덱스 초기화
groupped_reset = groupped.reset_index()
groupped_reset
```

```python
## 3. 다중 컬럼 pivot table 생성
pivot = df.pivot_table(index = '성별', columns =  ['흉통유형', '운동후 ST경사'], values = '최대 심박수', aggfunc = 'mean')
pivot
```

```python
## 4. 다중 컬럼 계층 전환
pivot_stack = pivot.stack(future_stack=True)
pivot_stack
```
