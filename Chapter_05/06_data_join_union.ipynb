```python
import pandas as pd

data1 = {
    '제품ID' : [1,2,3],
    '제품명' : ['노트북', '모니터', '키보드']
}

df1 = pd.DataFrame(data1)

data2= {
    '제품ID' : [1,2,4],
    '가격' : [1000, 200, 50]
}

df2 = pd.DataFrame(data2)

print(df1)
print(df2)
```

```python
# Data Join
##Inner Join 
inner_join = pd.merge(df1, df2, on = '제품ID', how = 'inner')
print(inner_join)
```

```python
## Outer Join
outer_join = pd.merge(df1, df2, on = '제품ID', how = 'outer')
print(outer_join)
```

```python
## Left Join
left_join = pd.merge(df1, df2, on = '제품ID', how = 'left')
print(left_join)
```

```python
## Right Join
right_join = pd.merge(df1, df2, on = '제품ID', how = 'right')
print(right_join)
```

```python
## Cross Join
cross_join = pd.merge(df1, df2, how = 'cross')
print(cross_join)
```

```python
# Data Union
## 결합 후 index 재설정
data1 = {
    '이름' : ['홍길동', '이순신'],
    '나이' : [30, 40]
}
df1 = pd.DataFrame(data1)

data2 = {
    '이름' : ['강감찬', '유관순'],
    '나이' : [50,20]
}
df2 = pd.DataFrame(data2)
```

```python
union_df = pd.concat([df1, df2], ignore_index = True)
print(union_df)
```

```python
## index 유지하며 union 
union_with_index = pd.concat([df1, df2], ignore_index = False)
print(union_with_index)
```

```python
data3 = {
    '이름' : ['정약용', '김유신'],
    '주소' : ['서울', '경주']
}
df3 = pd.DataFrame(data3)
```

```python
## dataFrame의 열이 다른 경우
union_with_different_columns = pd.concat([df1, df3], ignore_index = True)
print(union_with_different_columns)
```
