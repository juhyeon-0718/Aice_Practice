```python
import pandas as pd
path = 'C:/juehyeon/aice_test/materials/datasets/'
df = pd.read_csv(path + 'heart.csv')
```

```python
## 데이터 전처리 
# 최대 심박수 대비 나이 비율
df['heart_age_ratio'] = (df['thalachh']/ df['age']).round(2)
```

```python
# 누적합 : cumsum()
df['chol_cumsm'] = df['chol'].cumsum()
```

```python
# 10년 단위 그룹화 
df['age_group'] = (df['age'] // 10) * 10
```

```python
# 순위 : rank()
df['thalach_rank'] = df['thalachh'].rank()
```

```python
# 변화량 : diff()
df['chol_diff'] = df['chol'].diff()
```

```python
## 문자형 데이터 파생 함수
df = pd.DataFrame({
    'name' : ['김철수', '이영희', '박민수', '정수진', '이서준'],
    'email' : ['kim@naver.com', 'lee@gmail.com', 'partk@daum.net', 'jung@hanmail.net', 'seo@korea.kr'],
    'gender_eng' : ['Male', 'Female', 'Male', 'Female', 'Male']})
print(df)
                  
```

```python
# 문자열 길이 계산 str.len()
df['name_length'] = df['name'].str.len()
```

```python
# 대소문자 변환 : str.upper(), str.lower()
df['gender_upper'] = df['gender_eng'].str.upper()
df['gender_lower'] = df['gender_eng'].str.lower()
```

```python
# 특정 문자 포함 여부 확인 : str.contains('포함 문자')
df['has_수'] = df['name'].str.contains('수')
```

```python
# 문자열 바꾸기 : str.replace()
df['gender_kr'] = df['gender_eng'].str.replace('Male', '남성').str.replace('Female', '여성')
```

```python
# 문자열 자르기 : str.slice()
df['first_letter'] = df['name'].str.slice(0,1)
```

```python
# 문자열 나누기 : str.split().str[]
df['email_domain'] = df['email'].str.split('@').str[1]
```

```python
# 공백 제거 : str.strip()
df['name_stripped'] = df['name'].str.strip()
```

```python
# 문자열 연결 : + 
df['name_email_concat'] = df['name_stripped'] + '_' + df['email_domain']
```

```python
## 날짜형 데이터 파생 변수
df = pd.DataFrame({
    '가입일' : pd.to_datetime(['2022-01-15', '2023-03-10', '2023-12-01', '2025-05-03', '2024-10-20'])
})
```

```python
# 기준일 설정 
df['today'] = pd.to_datetime('2025-05-05')
```

```python
# 날짜 파생변수 생성
df['가입연도'] = df['가입일'].dt.year
df['가입월'] = df['가입일'].dt.month
df['가입일자'] = df['가입일'].dt.day
df['가입요일'] = df['가입일'].dt.day_name()
df['가입요일번호'] = df['가입일'].dt.weekday
df['가입_분기'] = df['가입일'].dt.quarter
df['가입_주차'] = df['가입일'].dt.isocalendar().week
df['월_시작일인가'] = df['가입일'].dt.is_month_start
df['월_마지막일인가'] = df['가입일'].dt.is_month_end
df['가입후_경과일수'] = (df['today'] - df['가입일']).dt.days 
```

```python
# 주말 여부 판단
df['주말가입여부'] = df['가입일'].isin([5,6])
```
