: 대표적인 머신러닝 라이브러리로, 간단한 api를 통해 데이터 전처리, 모델 학습 및 평가 수행 가능

   - 사용자 권한 설치: pip install scikit-learn --user
     
2) 라이브러리 계층 구조
   - library : 모든 기능 포함(ex: import sklearn)
   - Sub-library : 특정 작업이나 알고리즘 그룹을 위한 세부 모듈로 구성(ex: sklearn.model_selection, sklearn.linear_model etc)
   - Module : 서브라이브러리에 속한 세부 클래스나 함수 포함(ex: sklearn.linear_model.LinearRegression etc) 

```python
# 데이터 분리 : sklearn의 model_selection sub library에서 train_test_split module 가져오기
from sklearn.model_selection import train_test_split
```

```python
# 선형 회귀 모델: sklearn의 linear_model sub library에서 LinearRegression module 가져오기
from sklearn.linear_model import LinearRegression
```

```python
# 데이터 평가 : sklearn의 metrics sub library에서 mean_squared_error module 가져오기 
from sklearn.metrics import mean_squared_error
```

```python
# 데이터셋 생성
import pandas as pd
data = {'Feature' : [1,2,3,4,5], 'Target': [3,6,9,12,15]}
df = pd.DataFrame(data)
df.head()
```

```python
# 입력 (X : Feature), 출력 (Y : Target) 데이터 분리
X = df[['Feature']] # 2차원 데이터로 만들기
y = df['Target'] # 1차원 배열로 저장

X_train, X_test, y_train, y_test = train_test_split(
    X, y, # 입출력 데이터 넣기
    test_size = 0.2, # 전체 데이터의 20%를 테스트 데이터로 사용
    random_state = 42 # 결과 재현을 위해 랜덤 시드 고정
)

# 선형 회귀 모델 생성 및 학습 : LinearRegression
model = LinearRegression() #모델 객체 생성
model.fit(X_train, y_train) # 모델을 학습용 데이터로 학습 

# 학습된 모델로 테스트 데이터 예측 
predictions = model.predict(X_test)

# 모델 성능 평가
mse = mean_squared_error(y_test, predictions) # 평균 제곱 오차 계산
print("Mean Squared Error: ", mse)
    
```
