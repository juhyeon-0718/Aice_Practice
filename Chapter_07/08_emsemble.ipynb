# Emsemble (앙상블) 모델
: 여러 개의 머신러닝 모델(개별 모델)을 결합하여 하나의 강력한 모델을 만드는 기법
  개별 모델은 "약한 모델"일 수 있으나, 이를 결합함으로 성능을 개선할 수 있음
  주로 분류(Classification)와 회귀(Regression)문제에 사용

1) 앙상블 모델 종류
   - Voting (보팅) : 서로 다른 유형의 모델을 결합하여 결과를 결정
     ex) 로지스틱 + 의사결정나무 + 서포트 벡터 머신
   - Bagging (배깅) : 동일한 유형의 모델을 여러 번 학습시켜 결과를 결합
     ex) 10개의 의사결정 나무를 학습
   - Boosting (부스팅) : 약한 모델(Weak Leaner)을 순차적으로 학습하며 오류 보완

### Voting
    Soft Voting은 전문가들이 "찬성 확률 80%, 반대 확률 20%"처럼 세부적 의견을 제공하며, Hard Voting은 전문가들이 "찬성" 또는 "반대"로 의견을 표현하는 것과 같이 비유할 수 있음
    여러 개의 머신러닝 모델을 결합하여 결과를 결정하는 앙상블 방법, 주로 분류(Classifier) 문제에 활용
    - Hard Voting : 각 모델의 예측 결과를 다수결 투표로 결합하여 최종 결과를 결정, 구현이 직관적이며 계산이 빠름
    - Soft Voting : 각 모델이 예측한 화귤 값을 평균 내어 가장 높은 확률의 클래스를 선택, 더 세밀한 결과를 얻을 수 있음


    - 장점 : 다양한 모델 결합, 성능 향상
    - 단점 : 개별 모델 성능에 의존, 계산 비용 증가(여러 개 모델 결합하기 때문에)

### Bagging 
    "여러 요리사가 각각의 요리를 만들고, 가장 좋은 요리를 결정"
    동일한 모델을 여러 번 학습시켜 각각의 결과를 결합해서 최종 결과를 만듦, 데이터 샘플링 과정에서 중복 허용하여 모델들이 서로 다른 데이터에 대해 학습하도록 만듦
    - 과정 : 데이터 샘플링 - 모델 학습 - 결과 결합
    - 장점 : 과적합 방지 (각 모델이 서로 다른 데이터에 대해 학습하도록 하기때문에), 노이즈에 민감하지 않음 (여러 모델 결과를 결합하기 때문)
    - 단점 : 데이터가 적을 경우 샘플링 효과가감소, 학습 시간이 늘어남

### Boosting
    "학생이 반복적으로 시험을 보고, 틀린 문제를 집중적으로 공부하는 과정"
    - 약한 학습기를 순차적으로 학습하며, 이전 모델의 오류를 보완하는 방식. 
    - 주요 알고리즘 : AdaBoost(틀린 샘플에 가중치 부여하여 모델이 이를 더 잘 학습하도록), Gradient Boost(잔차를 최소화하는 방향으로 모델 순차적 학습)
    - 장점 : 성능이 뛰어남 (오류를 점진적으로 줄여가는 방식으로 작동), 데이터 불균형에 강함 (틀린 데이터에 가중치를 둬서 다음 모델이 이를 신경쓰도록 유도, 데이터 간 샘플 수 차이가 커도 틀린 데이터를 집중적으로 학습함으로써 소수 클래스에 대한 성능 향상 강함)
    - 단점 : 학습 시간이 오래 걸림, 과적합 위험 (학습 데이터를 과도하게 학습하거나 모델을 너무 학습할 경우)
