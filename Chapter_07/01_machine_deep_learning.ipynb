# Machine Learning (머신러닝) 
: 데이터를 학습하여 일정한 패턴을 발견하고 이를 기반으로 예측하거나 결정을 내리는 기술

1) 주요 특징
   - 데이터 중심적 학습 : 머신러닝은 데이터를 사용하여 모델을 학습시키기 때문에 데이터가 많고 품질이 좋을수록 모델 성능도 향상됨
   - 일반화 능력 : 학습한 데이터 외의 새로운 데이터에서도 높은 성능을 보이는 것을 목표로 함
   - 적응 능력: 학습 변화나 새로운 데이터에 대해 모델 재학습 가능
     
2) 머신러닝의 종류
   - 지도학습(Supervised Learning) : 입력 데이터와 함께 정답(라벨) 제공하며 이를 기반으로 모델 학습
     ex) 이메일의 스팸 여부 분리 
   - 비지도학습(Unsupervised Learning) : 정답이 없는 데이터를 기반으로 데이터 내 숨겨진 패턴 학습
     ex) 고객 세그먼트 군집화
   - 강화학습(Reinforcement Learning) : 에이전트가 환경과 상호작용하며 보상을 최대화하는 방법 학습
     ex) 게임 Ai, 로봇 제어

3) 머신러닝의 활용 예시 : 이미지 분류, 추천 시스템, 의료 진단

4) 동작 과정
   : 데이터 준비 - 모델 학습 - 예측 - 평가 및 개선

# Deep Learning (딥러닝)
: 머신러닝의 한 분야로, 인공신경망이라는 구조를 사용해 데이터를 학습하는 기술. 방대한 데이터와 복잡한 문제를 처리하는 데 강력한 성능 발휘하며 사람의 뇌를 모방하여 여러 층으로 구성된 신경망을 통해 학습

1) 주요 특징
   - 계층적 학습 구조 : 여러 개의 층을 통해 데이터를 점진적으로 학습
     ex) 첫 번째 층 - 이미지의 가장 기본적 특징, 두 번째 층 - 기본적 특징을 조합해 복잡한 모양 학습, 마지막 층 : 학습 내용을 바탕으로 이미지를 특정 사물로 분류
    - 대량의 데이터 학습 가능 : 데이터가 많아질수록 성능 향상
    - 비선형 관계 학습 : 단순히 선형 관계뿐만 아니라 비선형적 데이터 간의 관계도 학습 가능

2) 딥러닝의 종류
    - 합성공 신경망(CNN, Convolutional Neural Networks) : 이미지 처리에 특화된 딥러닝 모델 ex) 얼굴 인식, 자율 주행 차량
    - 순환 신경망(RNN, Recurrent Neural Networks) : 시간적 순서가 중요한 데이터 학습 ex) 음성인식, 텍스트 번역, 주가 예측
    - 생성적 적대 신경망(GAN, Generative Adversarial Networks) : 새로운 데이터를 생성하는 데 사용 ex) 가짜 이미지 생성, 예술작품 스타일 변환

3) 활용 예시 : 이미지 분석(X-ray 이미지 분석하여 질병 진단), 음성 인식(스마트폰이 사용자의 음성 인식하여 명령 수행), 자연어 처리(챗봇이 사용자의 질문에 적절한 답변 제공), 추천 시스템(넷플릭스에서 사용자가 좋아할만한 컨텐츠 제공)

4) 동작 과정 : 데이터 준비 - 신경망 설계 (입력층 / 은닉층 / 출력층) - 모델 학습 - 예측 - 평가 및 개선 

## 지도학습 

1) 작동원리
   - 데이터 준비 : 입력값(X)와 출력값(Y)로 나눔 
   - 모델 학습 : 입력(X) -> 출력(Y)의 관계 학습하여 데이터를 분석해 패턴과 규칙 찾아냄
   - 예측 : 학습한 모델을 사용하여 입력 데이터에 대한 결과를 예측
   - 평가 및 개선 : 모델의 예측 결과를 실제 정답과 비교하여 성능 평가. 모델이 부족한 점이 있으면 추가 학습이나 데이터 조정을 통해 개선
2) 활용 예시 : 이미지 분류, 스팸메일 분류, 질병 진단

## 비지도학습

1) 작동원리
   - 데이터 준비 : 비지도학습에서는 입력 데이터(X)만 사용 (ex: 대형 쇼핑몰의 고객 데이터 분석시 x는 고객의 나이, 구매 횟수, 소비 금액 등)
   - 데이터를 분석하여 유사한 데이터끼리 그룹화하거나 특징 추출
   - 결과 활용 : 발견한 패턴이나 그룹을 활용하여 데이터 분석하거나 의사결정에 활용
2) 활용 예시 : 고객 세그먼트 분류(고가 제품 선호, 할인 제품 선호 고객 등), 이상치 탐지(금융 사기 탐지), 문서 군집화(뉴스 기사 자동 분류)

## 지도 vs. 비지도 학습
| 구분 | 지도학습(Supervised Learning) | 비지도학습(Unsupervised Learning) |
| :--- | :--- | :--- |
| **라벨** | 정답(라벨)이 필요함 | 정답(라벨)이 필요 없음 |
| **목표** | 입력과 출력 간 관계 학습 | 데이터 간 패턴이나 구조 발견 |
| **활용** | 예측 문제(분류, 회귀) 해결 | 데이터 분석 및 그룹화, 패턴 탐지 |
| **예시** | 이메일 스팸 분류, 질병 진단 | 고객 세그먼트 분류, 이상치 탐지 |
