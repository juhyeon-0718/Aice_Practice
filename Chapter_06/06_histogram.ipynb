```python
import pandas as pd
import matplotlib.pyplot as plt

path = 'C:/juehyeon/aice_test/materials/datasets/'
data = pd.read_csv(path + 'heart.csv')
data.head()
```

```python
# 히스토그램 생성
plt.hist(
    data['age'], # (data의 age 열 사용)
    bins = 10, # 구간 수 (10개로 나눔) 
    color = 'skyblue', # 막대 색
    alpha = 0.7, # 투명도 
    edgecolor = 'black', # 테두리 색 
    # density = True # 정규화 여부 
)
plt.title("Age Destribution in Heart Dataset")
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.grid(True) # 격자 추가 
plt.show()
```
