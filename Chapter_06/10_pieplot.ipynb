# 파이 그래프 (Pie plot)
: 파이 그래프는 데이터의 비율을 시각화하는 데 사용, 전체 데이터에서 각 항목이 차지하는 비율 직관적으로 보여줌

## 구성요소
1) 원형 : 데이터의 전체 크기
2) 조각 : 각 항목의 비율

## 활용 사례 
1) 시장점유율 분석
2) 예산 항목별 비중 비교
3) 고객 세그먼트 비율 시각화

```python
import matplotlib.pyplot as plt
```

```python
categories = ['Category A', 'Category B', 'Category C', 'Category D']
values = [30, 20, 35, 15]
```

```python
plt.pie(
    values, 
    labels = categories, # 항목 이름 설정
    autopct = '%1.1f%%', # 비율 표시 형식
    explode = (0, 0.1, 0, 0), # 특정 항목 강조 
    colors = ['skyblue', 'orange', 'green', 'red'], # 섹션 색상 
    shadow = True # 그림자 추가
)

plt.title('Sample Pie Chart')
plt.show()
```

```python
# 여러 개의 파이그래프를 하나의 그림에 배치하기

## 두 개의 그래프를 위한 subplot 생성
flg, axs = plt.subplots(1,2) 

## 첫 번째 파이 그래프 
axs[0].pie(
    values, 
    labels = categories,
    autopct = '%1.1f%%',
    colors = ['skyblue', 'orange', 'green', 'red']
)

## 두 번째 파이 그래프
axs[1].pie(
    values,
    explode  = (0, 0.2, 0, 0),
    labels = categories,
    autopct = '%1.1f%%'
)
plt.show()
```

### 추가 옵션

| 옵션 | 설명 | 예제 |
| :--- | :--- | :--- |
| **labels** | 각 항목의 이름을 설정합니다. | `labels=['A', 'B', 'C']` |
| **autopct** | 각 항목의 비율 표시 형식을 지정합니다. | `autopct='%1.1f%%'`, `autopct='%d%%'` |
| **startangle** | 그래프의 시작 각도를 설정합니다. | `startangle=90` |
| **explode** | 특정 항목을 강조하여 원에서 분리합니다. | `explode=(0, 0.1, 0, 0)` |
| **colors** | 각 섹션의 색상을 설정합니다. | `colors=['red', 'blue', 'green']` |
| **shadow** | 그래프에 그림자를 추가합니다. | `shadow=True` |
