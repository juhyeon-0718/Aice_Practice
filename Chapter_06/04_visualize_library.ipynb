# 시각화 
    : 데이터를 쉽게 이해하고 분석할 수 있도록 시각적으료 표현하는 과정

**시각화 라이브러리 종류** 
1. Matplotlib : 기초적 그래프 작성에 적합하며 커스터마이징 가능
2. Seaborn : 통계적 데이터 시각화를 위한 고급 그래프 제공
3. Plotly : 대화형 시각화 지원. HTML로 내보내기 가능
4. Altair : 선언형 문법 사용하여 간결하게 시각화 가능

```python
# Matplotlib 
import matplotlib.pyplot as plt
from matplotlib import rc
```

```python
rc('font', family = 'Malgun Gothic')
```

```python
x = [1,2,3,4,5]
y = [10,20,30,40,50]
```

```python
plt.plot(x, y, label = "한글데이터", color = 'blue', marker = 'o')
plt.title('한글 데이터 테스트')
plt.xlabel('x축')
plt.ylabel('y축')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Seaborn
import seaborn as sns
import pandas as pd

data = pd.DataFrame({
    'Category' : ['A', 'B', 'C', 'D'],
    'Values' : [5, 7, 3, 8]
})


sns.barplot(x = 'Category', y = 'Values', data = data)
plt.title("Bar Plot Example")
plt.show()
```

```python
# Plotly 
import plotly.express as px 

data = px.data.gapminder()
# data.head()
fig = px.scatter(
    data, 
    x = 'gdpPercap', # x축에 사용할 열 지정
    y = 'lifeExp', # y추에 사용할 열 지정
    size = 'pop', # 점 크기를 나타낼 열 지정
    color = 'continent', # 점 색상 결정할 열 지정
    log_x = True, # log scale 사용 여부 지정 (*log scale : 데이터가 큰 범위로 분포되어 있을 때 압축)
    title = "Interactive Scatter Plot" 
)

fig.update_layout(width = 800, height = 500)  # 전반적 레이아웃 조정
fig.show()
```

```python
 # Pandas Plot :  데이터 프레임 조작에 맞춰진 라이브러리지만 간단한 시각화 기능도 제공
data = {'Year' : [2020, 2021, 2022], 'Sales' : [250, 300, 400]}
df = pd.DataFrame(data)
df.plot(x = 'Year', y = 'Sales', kind = 'line', title = "Sales Over Year")
plt.show()
```
