# Line Plot 
: 시간의 흐름에 따른 데이터 변화를 시각화하는 데 사용. 데이터의 연속적 변화를 나타내며, 시간에 따른 추세를 파악하는 데 적합

## 구성 요소
1) X축 : 시간 또는 순서를 나타냄
2) Y축 : 시간에 따른 값의 변화를 나타냄
3) 선 : 데이터가 시간에 따라 어떻게 변화했는지를 나타냄

## 활용 사례
1) 주식 시장에서 시간별 가격 변화를 시각화
2) 월별 매출 추세 분석
3) 온도 변화 데이터 분석

```python
import pandas as pd
import matplotlib.pyplot as plt

path = 'C:/juehyeon/aice_test/materials/datasets/'
data = pd.read_csv(path + 'heart.csv')
data.head()
```

```python
### 데이터 정렬하기
data = data.sort_values(by = "age")
data.head()
```

```python
### 선 그래프 생성
plt.plot(
    data['age'], # x축
    data['chol'], # y축
    color='blue', # 선 색깔 
    linewidth = 2, # 선 두께 
    linestyle = '-', # 실선
    marker = 'o' ,#데이터 점 표시
    markersize = 5, # 점 크기 
    label = 'Cholesterol' # 범례
)

### 그래프 꾸미기
plt.title("Age vs. Cholesterol")
plt.xlabel('age')
plt.ylabel('Cholesterol')
plt.legend() # 범례 표시 
plt.grid(True)
plt.show()
```

### 추가 옵션

| 옵션 | 설명 | 예제 |
| :--- | :--- | :--- |
| **color** | 선 색상을 지정합니다. | `color='blue'`, `color='#FF5733'` |
| **linewidth** | 선의 두께를 설정합니다. | `linewidth=2` |
| **linestyle** | 선 스타일을 설정합니다. | `'-'`(실선), `'--'`(점선) |
| **marker** | 데이터 지점을 표시할 마커 스타일을 설정합니다. | `'o'`(원), `'x'`, `'^'`(삼각형) |
| **markersize** | 마커의 크기를 설정합니다. | `markersize=5` |
| **label** | 범례에 표시될 레이블을 설정합니다. | `label='Cholesterol'` |
