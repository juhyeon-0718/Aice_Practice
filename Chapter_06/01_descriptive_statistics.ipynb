```python
import pandas as pd
data = {'Score' : [80, 90, 100, 70, 85]}
df = pd.DataFrame(data)
```

```python
# 평균값 : .mean()
mean_value = df['Score'].mean()
print(mean_value)
```

```python
# 중앙값 : .median()
median_value = df['Score'].median()
print(median_value)
```

```python
# 최빈값 : .mode() **데이터에 중복이 없을 경우 여러개의 최빈값이 나올 수 있음**
mode_value = df['Score'].mode()
print(mode_value)
```

```python
# 최댓값과 최솟값: .min(), .max()
min_value = df['Score'].min()
max_value = df['Score'].max()

print(min_value)
print(max_value)
```

```python
# 표준현차 : .std()
std_value = df['Score'].std()
print(std_value)
```

```python
# 분산 : .var()
variance_value = df['Score'].var()
print(variance_value)
```
