# Bar plot (막대 그래프)
: 서로 다른 그룹 간의 데이터를 비교하는 데 적합하며, 각 데이터의 크기를 막대의 길이로 표현

## 구성요소
1) x축 : 비교 대상 (제품, 지역, 그룹)
2) y축 : 해당 항목의 데이터 값
3) 막대 : 데이터 값을 시각적으로 나타냄

## 활용사례
1) 제품별 매출 데이터 비교
2) 연도별 성장률 분석
3) 고객 선호 제품 유형 비교 

```python
import pandas as pd
import matplotlib.pyplot as plt

path = 'C:/juehyeon/aice_test/materials/datasets/'
data = pd.read_csv(path + 'heart.csv')

data.head()
```

```python
age_groups = data.groupby('age')['chol'].mean().reset_index()
```

```python
##  막대 그래프 생성
plt.bar(
    age_groups['age'],
    age_groups['chol'],
    color = 'skyblue',
    width = 0.8,
    edgecolor = 'black'
)

plt.title("Average Cholesterol by Age")
plt.xlabel('Age')
plt.ylabel('Cholesterol')
plt.xticks(rotation = 45) # x축 레이블 회전
plt.grid(axis = 'y', linestyle = '-')
plt.show()
```

```python
## 수평 막대 그래프 생성 : plt.barh()
plt.barh(
    age_groups['age'],
    age_groups['chol'],
    color = 'orange',
    edgecolor = 'black'
)

plt.title("Average Cholesterol by Age(Horizontal)")
plt.xlabel("Average Cholesterol")
plt.ylabel("Age")
plt.show()
```

### 추가 옵션

| 옵션 | 설명 | 예제 |
| :--- | :--- | :--- |
| **color** | 막대의 색상을 설정합니다. | `color='skyblue'`, `color='#FF5733'` |
| **width** | 막대의 너비를 설정합니다. | `width=0.8` |
| **edgecolor** | 막대 테두리 색상을 설정합니다. | `edgecolor='black'` |
| **align** | 막대의 위치를 정렬합니다. | `'center'`(기본값), `'edge'` |
| **tick_label** | X축 레이블을 사용자 지정합니다. | `tick_label=['A', 'B', 'C']` |
| **label** | 범례에 표시될 레이블을 설정합니다. | `label='Average Cholesterol'` |
