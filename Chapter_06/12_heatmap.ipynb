# 히트맵 (Heatmap) 
: 데이터의 상관관계나 패턴을 색상으로 표현하는 그래프. 대규모 데이터를 요약하거나, 변수 간의 상관성을 직관적으로 나타낼 때 사용

## 구성요소
1) 행과 열: 분석하려는 변수
2) 셀 : 각 변수 간의 관계를 색상으로 표현

## 활용 사례 
: 변수 간 상관관계 분석, 지역별 사고 빈도 분석, 데이터 밀도 시각화

```python
#  최대 심박수 (thalachh)와 콜레스테롤 (chol) 값 간의 관계를 산점도로 표현
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

path = 'C:/juehyeon/aice_test/materials/datasets/'
data = pd.read_csv(path + 'heart.csv')
data.head()
```

```python
## 상관관계 계산 : .corr()
correlation_matrix = data.corr()

## 히트맵 생성
plt.figure(figsize = (10,8))
sns.heatmap(
    correlation_matrix,
    annot = True,
    fmt = ".2f",
    cmap = "coolwarm",
    linewidths = 0.5,
    cbar = True
)

plt.title("Correlation Heatmap")
plt.show()
```

### 추가 옵션

| 옵션 | 설명 | 예제 |
| :--- | :--- | :--- |
| **annot** | 각 셀에 값을 표시합니다. | `annot=True` |
| **fmt** | 셀에 표시되는 숫자의 형식을 지정합니다. | `fmt=".2f"` |
| **cmap** | 색상 맵을 설정합니다. | `cmap="coolwarm"`, `cmap="viridis"` |
| **linewidths** | 셀 간의 경계선 두께를 설정합니다. | `linewidths=0.5` |
| **cbar** | 컬러바를 표시하거나 숨깁니다. | `cbar=False` |
