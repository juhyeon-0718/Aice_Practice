## 시각화 기본 흐름

1) **x축 변수** : 기준이 되는 항목 -> 연속형 데이터, 범주형 데이터
2) **y축 변수** : 수치형 결과 -> 평균, 합계, 개수 etc
3) **그래프 형태 선택** : 선 / 막대 / 산점도 etc
4) **제목, 축 이름** 
5) **그래프 출력**

```python
# Seaborn 
import seaborn as sns
# 한글 포늩 지정
plt.rcParams['font.family'] = "Malgun Gothic"

# 마이너스 부호 깨짐 방지
plt.rcParams['axes.unicode_minus'] = False
```

```python
## line graph
sns.lineplot(data = df, x= 'age', y = 'chol')
plt.title("나이에 따른 콜레스테롤 변화")
plt.show()
```

```python
## 막대 그래프
sns.barplot(data = df , x = 'sex', y = 'chol')
plt.title('성별에 따른 평균 콜레스테롤')
plt.show()
```

```python
## scatter graph 
sns.scatterplot(data = df, x = 'age', y = 'thalachh')
plt.title('나이와 최대 심박수 산점도')
plt.show()
```

```python
## histogram 
sns.histplot(data = df, x = 'age', bins = 10, kde = True) # bins : 구간 수, kde : 커널 밀도 추정 (선)
plt.title("나이분포")
plt.show()
```

```python
## box graph
sns.boxplot(data = df, x = 'sex', y = 'thalachh')
plt.title('성별에 따른 심박수 분포 (이상치 포함)')
plt.show()
```

### 시각화 5단계 

| 단계 | 설명 | 대표 코드 예시 |
| :--- | :--- | :--- |
| ① 그래프 틀 선택 | 어떤 그래프를 그릴지 선택 | `plt.plot()`, `plt.bar()`, `sns.scatterplot()` |
| ② 데이터 넣기 | x축, y축에 어떤 변수를 넣을지 결정 | `x=df['age']`, `y=df['chol']` |
| ③ 이름 붙이기 | 축 이름, 제목 작성 | `plt.xlabel()`, `plt.title()` |
| ④ 꾸미기 | 색상, 선 모양, 눈금, 스타일 | `color=`, `linestyle=`, `grid=True` |
| ⑤ 출력 | 그래프를 화면에 출력 | `plt.show()` |

### seaborn 함수별 옵션 요약

| 함수 | 주요 옵션 | 설명 |
| :--- | :--- | :--- |
| `sns.lineplot()` | x, y, data, hue, style | 선그래프, 그룹별 색/선 모양 구분 |
| `sns.barplot()` | x, y, data, est, ci | 평균 또는 합계를 막대로 표현 |
| `sns.scatterplot()` | x, y, data, hue, size, alpha | 그룹별 산점도, 점 크기/투명도 |
| `sns.histplot()` | x, bins, kde | 분포 시각화 (히스토그램) |
| `sns.boxplot()` | x, y, data | 이상치 포함한 사분위 시각화 |
| `palette` | 'pastel', 'deep' 등 | 컬러 테마 지정 |
| `hue, style, size` | 컬러명 | 그룹별 색상, 선스타일, 점크기 |
| `plt.legend()` | loc='upper right' 등 | 범례 위치 조정 |

### matplotlib.pyplot 함수별 옵션 요약

| 함수 | 주요 옵션 | 설명 |
| :--- | :--- | :--- |
| `plt.plot()` | color, linestyle, marker, linewidth | 선 그래프 색, 점 모양, 선 스타일 설정 |
| `plt.bar()` | color, width | 막대 색상과 너비 설정 |
| `plt.scatter()` | color, s, alpha | 산점도 색상, 점 크기, 투명도 |
| `plt.xlabel()`, `plt.ylabel()` | '텍스트' | 축 이름 |
| `plt.title()` | '텍스트' | 그래프 제목 |
