# 산점도 (Scatter plot)
: 두 변수 간의 관계를 시각적으로 표현하는 데 사용. 데이터의 분포와 패턴을 확인하고 상관관계 파악 가능

## 구성요소 
1) x축 : 독립변수 (광고비)
2) y축 : 종속변수 (매출액)
3) 점 : 데이터의 개별 관측값

## 활용 사례 
: 광고비와 매출 간의 상관관계 분석, 키와 몸무게 간의 관계 확인, 학습 시간과 시험 점수 간의 패턴 분석

```python
# 최대 심박수 (thalach)와 콜레스테롤 (chol) 값 간의 관계를 산점도로 파악하기
import pandas as pd
import matplotlib.pyplot as plt

path = 'C:/juehyeon/aice_test/materials/datasets/'
data = pd.read_csv(path + 'heart.csv')
data.head()
```

```python
## 산점도 생성
plt.scatter(
    data['chol'],
    data['thalachh'],
    c = data['age'], # 점 색상 : 나이 
    s = 50, # 점 크기
    alpha = 0.7, # 점 투명도
    cmap = 'viridis', # 색상 맵
    edgecolor = 'black' # 점 테두리 색상
)

plt.title("Cholesterol vs Max Heart Rate")
plt.xlabel("Cholesterol")
plt.ylabel("Max heart Rate")
plt.colorbar(label = "Age")
plt.grid(True)
plt.show()
    
```
