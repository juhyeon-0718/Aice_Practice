`tensorflow.keras`를 이용하여 회귀 모델을 만드세요.

- 세 번째 Hidden Layer: `32`개 뉴런, `relu`

## 5-14. 세 모델의 성능 평가

최종적으로 다음 세 모델을 비교합니다.

- `ridge_model`
- `xgb_model`
- `model_es`

- `xgb_pred`

각 모델에 대해 다음 두 평가 지표를 계산하세요.

- `R²`
- `MAE`

사용 함수:

- `mean_absolute_error`

- **R² → 클수록 좋음**
- **MAE → 작을수록 좋음**

두 지표가 서로 다른 모델을 가리키는 경우에는 결과를 모두 비교하고 이유를 설명하세요.
