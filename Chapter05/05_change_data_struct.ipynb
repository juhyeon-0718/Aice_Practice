```python
import pandas as pd
path = 'C:/juehyeon/aice_test/materials/datasets/'
df = pd.read_csv(path + 'heart.csv')
```

```python
df.columns = ['나이', '성별', '흉통유형', '안정시혈압', '콜레스테롤', '공복혈당', '심전도', '최대 심박수',
              '운동유발협심증', 'ST 우울증', '운동후 ST경사', '주요혈관수', '결함유형', '심장질환여부']
```

```python
# 그룹화와 집계 
## 성별을 기준으로 데이터 그룹화
groupped= df.groupby('성별')
groupped.size()
```

```python
## 성별별 주요 열의 평균 집계
groupped_mean = groupped.mean()
print(groupped_mean)
```

```python
## 여러 집계 함수 사용 : agg()
### agg() : 여러 열에 다양한 집계 함수 동시 적용, 딕셔너리 형태로 열 이름과 함수 목록 지정
groupped_stats = groupped.agg({
    '나이' : ['max' , 'min'],
    '콜레스테롤' : ['mean', 'std']
})
print(groupped_stats)
```

```python
## 다중 그룹화
groupped_multi = df.groupby(['성별', '흉통유형']).mean()
groupped_multi
```

```python
# Pivot Table 조작 : 특정 기준에 따라 데이터를 요약하거나 변환하는데 사용하는 도구
## 1. pivot table 생성
pivot = df.pivot_table(index = '성별', columns = '흉통유형', values = '최대 심박수', aggfunc = 'mean')
### 해석 ) 성별을 인덱스로 지정, 흉통유형을 열로 지정, 최대 심박수 값을 피벗 테이블에서 요약 데이터로 사용, 최대 심박수의 평균값 계산
pivot
```
