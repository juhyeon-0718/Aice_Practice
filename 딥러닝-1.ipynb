# 딥러닝

: 인공지능의 핵심 기술, 인간의 뇌 신경망 구조에서 영감을 받아 설계된 알고리즘을 이용하여 데이터를 처리하고 복잡한 패턴 학습하는 기술. 데이터를 기반ㄹ으로 가중치와 편향을 학습하고 이를 통해 데이터의 중요한 특징을 자동으로 추출. 

## 특징 
1. 자동 특징 추출: 데이터를 기반으로 복잡한 특징을 자동으로 학습하여 전처리 과정 줄여줌
2. 확장 가능성 : 대량의 데이터를 학습하여 모델 성능을 점진적으로 향상 시킬 수 있음
3. 다양한 구조 : 합성곱 신경망(cnn), 순환 신경망(rnn), 트랜스포머 등 문제 유형에 적합한 다양한 신경망 구조 설계 가능


## 1. 인공 신경망

: 인공신경망 (Artificial Neural Network, ANN)은 인간의 뇌에서 신경세포가 신호를 주고 받는 과정을 모방하여 설계된 기계 학습 모델. 데이터를 입력받아 학습을 통해 입력과 출력 간 관계를 모델링하며 복잡한 문제 해결시 활용 
1) 입력층
   - 역할 : 원시 데이터를 받아들임
   - 특징 : 데이터의 차원 수에 따라 뉴런 결정, 데이터를 신경망으로 전달하는 시작점 역할을 수행
2) 은닉층
   - 역할 : 중간 표현을 학습
   - 특징 : 입력 데이터의 특징을 추출하며, 여러 층으로 구성된 경우 신경망의 복잡성, 성능 향상
3) 출력층
   - 역할 : 예측값 반환
   - 출력 데이터 형태에 따라 뉴런 개수와 활성화 함수 달라짐
- 뉴런 : 인공신경망에서 데이터를 처리하는 기본 단위, 입력값을 받아 편항 계산 후 활성화 함수를 통해 결과 생성
- 노드 : 뉴런이 위치하는 구조적 단위, 신경망 내에서 데이터가 흐르고 처리되는 실제적 공간 제공
- 가중치 : 뉴런 간 연결의 강도 또는 중요도를 나타내는 값. 데이터의 특정 특징이 얼마나 중요한지 결정, 최적화 알고리즘을 통해 지속적 조정됨
- 편향 : 뉴런의 계산 결과에 추가로 더해지는 값. 유연한 데이터 학습을 도와주며 모든 입력값이 0일 때도 출력값이 특정 기준값을 가질 수 있도록 조정하는 역할.
4) 활성화 함수: 뉴런의 출력값을 비선형적으로 변환하는 역할. 단순 연산이 아닌 복잡한 패턴과 관계를 학습할 수 있도록 함. (인공신경망이 선형 관계만 학습한다면 층을 아무리 추가해도 복잡한 데이터를 표현할 수 없음)
      - 시그모이드 함수 : S자 형태의 곡선을 가진 비선형 활성화 함수. 입력값을 0과 1사이로 변환, 출력 값이 확률처럼 해석될 수 있어 이진 분류에서 자주 사용
      - 하이퍼볼릭 탄젠트 : 시그모이드와 유사한 S자 형태이지만, 출력 범위가 -1에서 1사이임. 시그모이드의 **출력 비중심화**(출력이 0과 1사이에 머물러 평균이 0에서 멀어짐) 문제를 개선
      - 렐루 함수: 딥러닝에서 널리 사용되는 활성화 함수, 입력값이 양수 -> 그대로 반환, 음수 -> 0으로 만들어줌 . 기울기 소실 문제를 해결할 수 있지만 음수일 경우 0으로 출력되기 때문에 죽은 ReLU 문제 발생 가능, 입력값이 너무 크면 기울기 폭주 문제 발생 가능
      - 리키 렐루 : 렐루 함수의 단점을 개선한 활성화 함수, 음수일 때도 작은 기울기를 갖도록 설계함.
      - 소프트맥스 : 신경망에서 출력값을 확률로 변환하는 함수. 주로 다중 클래스 분류에서 마지막 출력층 활성화 함수로 사용. 입력값을 0~1사이의 값으로 변환하여 이 값을 확률로 해석할 수 있도록 정규화하는 함수. 모델이 출력한 각 클래스의 점수를 전체 클래스 점수의 합과 비교하여 상대적 확률을 계산함으로써 모든 클래스의 확률 합이 항상 1이되도록 보장함. 

## 2. 심층신경망

: 다층 구조로 이루어진 인공 신경망, 입력 데이터에서 복잡한 패턴과 특징을 학습하여 예측이나 분류와 같은 작업을 수행함. 여러개의 은닉층을 포함하여 각각의 층이 입력 데이터를 점차 추상화하며 더 높은 수준의 특징을 추출

[관련 용어]
- 샘플: 하나의 데이터 단위
- 입력: 모델에 주어지는 데이터
- 타겟: 모델이 예측해야 하는 정답
- 모델: 데이터를 처리해 예측값을 생성하는 딥러닝 구조
- 파라미터: 학습 과정에서 업데이트 되는 값
- 손실함수: 예측값과 실제값의 차이를 측정하는 함수
- 손실: 손실 함수를 계산한 결과값. 낮을수록 모델 성능이 좋음
- 학습률: 가중치를 업데이트 할 때 변화의 크기를 결정하는 하이퍼파라미터
- 에폭: 전체 데이터셋이 모델을 통해 한 번 학습되는 과정
- 경사하강법: 손실을 최소화하기 위해 가중치를 업데이트 하는 최적화 알고리즘
- 오차역전파: 손실에 따라 파라미터 기울기 계산하고 가중치를 조정하는 과정
- 드롭아웃: 학습 중 일부 뉴런을 제외해 과적합 방지하는 기법
- 정규화: 모델 복잡도를 제어하여 일반화 성능을 높이는 기법
